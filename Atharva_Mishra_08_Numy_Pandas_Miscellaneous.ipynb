```python
np.random.seed(42)

marks = np.random.randint(20, 101, size=(12, 4))
print("Student Marks:")
print(marks)
student_avg = np.mean(marks, axis=1)
print("Student Averages:")
print(student_avg)
test_mean = np.mean(marks, axis=0)
print("Test Mean:")
print(test_mean)
test_std = np.std(marks, axis=0)
print("Test Standard Deviation:")
print(test_std)
passed = np.all(marks >= 40, axis=1)
print("Students passing all tests:")
print(np.where(passed)[0] + 1)
high_avg = np.where(student_avg >= 75)[0] + 1
print("Students with average >= 75:")
print(high_avg)
max_variation = np.argmax(test_std) + 1
print("Test with greatest variation:")
print(max_variation)
```

```python
np.random.seed(42)
data = np.random.normal(50, 12, 100)
print(data)
```

```python
cleaned = np.where(
    data < 30, 30,
    np.where(data > 70, 70, data)
)

print(cleaned)
```

```python
print("Original Mean:", data.mean())
print("Original Variance:", data.var())
print("Original Standard Deviation:", data.std())
print("Original Minimum:", data.min())
print("Original Maximum:", data.max())
```

```python
print("Cleaned Mean:", cleaned.mean())
print("Cleaned Variance:", cleaned.var())
print("Cleaned Standard Deviation:", cleaned.std())
print("Cleaned Minimum:", cleaned.min())
print("Cleaned Maximum:", cleaned.max())
```
