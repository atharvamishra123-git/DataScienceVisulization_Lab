```python
from pandas import Series
stock = Series([25, 40, 18], index=['Item1', 'Item2', 'Item3'])

stock_dict = stock.to_dict()
print(stock_dict)

stock2 = Series(stock_dict)
print(stock2)
```

```python
sales = Series(
    [650, 400, 800, 300],
    index=['Laptop', 'Mouse', 'Keyboard', 'Monitor']
)
print(sales.values)
print(sales.index)
print(sales['Laptop'])
print(sales[sales > 500])
print('Mouse' in sales.index)
sales_dict = sales.to_dict()
print(sales_dict)
```

```python
data = {'A': 15, 'B': 35, 'C': 25}

s = pd.Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'

print(s)
print(s[s > 20])
print(pd.notnull(s))
print(s.name)
print(s.index.name)
```
