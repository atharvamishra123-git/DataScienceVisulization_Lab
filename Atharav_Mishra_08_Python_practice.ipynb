```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(5))
print(my_abs(-7))
print(my_abs(0))
```

```python
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
```

```python
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:")
    print(val1)
    print(val2)

x = 6
y = 3

swap_message(x, y)

print("Outside function:")
print(x)
print(y)
```
